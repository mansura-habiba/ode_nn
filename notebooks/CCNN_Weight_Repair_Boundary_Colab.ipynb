# When Is CCNN Weight-Space Repair Actually Useful?
### A constrained model-editing study with fault-specific recovery, source preservation and representation-aware negative controls

**Research programme:** *CCNN structural memory × attention × weight-space repair* · **Evidence status: NOT RUN** (this file contains no invented measured outcomes)

**One research question.** After a trained attention–CCNN network suffers a genuine model-weight fault, **under which types of failure does its low-dimensional lag-kernel representation allow better targeted recovery and lower collateral damage than tangent, unrestricted-filter and ordinary-optimizer alternatives?**

**Scope:** Weight-space repair is the main endpoint of *this companion study*. It does not assume that CCNN improves prediction or attention in general. The clean network must first demonstrate nontrivial recall and use its memory branch. We investigate a *conditional* repair advantage, not a claim to beat all editing algorithms.

**Abstract — preregistered experiment.** We adapt the original *Zoology* multi-query associative-recall generator (Arora et al., ICLR 2024; exact source pinned), train an attention–CCNN network, and evaluate controlled parameter damage on the saved checkpoint. We compare **five-dimensional nonlinear CCNN edits, five-dimensional CCNN tangent edits, five-dimensional unrestricted FIR edits, full FIR edits, five-coordinate sparse Adam, and full-model Adam**, with zero-fault sham editing and a no-edit control. Faults include interpretable delay/bandwidth/query-dependence changes and off-manifold weight corruption. For each method we report calibration-detectable fault headroom, locked-test recovery, untouched-query preservation, placebo-adjusted excess, edited scalars, and compute. All faults, including failures to calibrate, remain in the output. The analysis maps *what works, what fails and the hypothesized representation boundary*; explanatory claims must be based on observed results.

## 1. Source basis, new question and hypotheses

**Original CCNN:** M. Habiba and B. A. Pearlmutter, *Continuous Convolutional Neural Networks: Coupled Neural PDE and ODE* (2021), [arXiv:2111.00343](https://arxiv.org/abs/2111.00343), describes a time-varying causal lag kernel coupled to a dynamical state. This study uses its **finite-lag kernel/PDE-accumulation plus leaky global-state approximation** inside a proposed attention-logit coupling. This is an adaptation, not the original paper's attention model.

**Benchmark:** [Zoology: Measuring and Improving Recall in Efficient Language Models](https://arxiv.org/abs/2312.04927), ICLR 2024. We load the **exact pinned upstream MQAR generator**; no CCNN kernel determines the labels. The experiment uses a restricted published-condition *64-token, four-pair* profile in the pilot; that is not the whole ICLR experiment.

**Model-editing precedents:** [AlphaEdit](https://proceedings.iclr.cc/paper_files/paper/2025/hash/29c8c615b3187ee995029284702d3f43-Abstract-Conference.html) (ICLR 2025) motivates explicit unrelated-capability retention; [CrispEdit](https://proceedings.mlr.press/v306/ikram26a.html) (ICML 2026) motivates preservation-constrained optimization and strong numerical controls. **We do not implement or claim to reproduce their algorithms.** The emerging weight-space-programming literature motivates structural editability, but these gradient-based experiments cannot establish LLM reasoning about weights.

**H-R1 (mechanism-localized repair):** for a model in which CCNN memory contributes to recall, damage to interpretable kernel mechanisms can be repaired using a five-dimensional nonlinear structural edit with equal or better recovery–preservation efficiency than an unrestricted five-dimensional FIR edit.

**H-R2 (nonlinear geometry):** under sufficiently large *detectable* on-manifold faults, nonlinear structured edits may outperform the same five-dimensional first-order tangent edit; for small faults, tangent should be competitive.

**H-R3 (boundary):** when corruption is outside the CCNN kernel's editable subspace (e.g. signal projection), structured edits should lose to a less restricted or full-model optimizer. This is a planned *negative control*—not a way to exclude inconvenient failures.

**Key null:** sparse Adam tuning the *same five CCNN parameters* should not be credited as a novel structural editor; if it matches the proposed edit, we have demonstrated an editable parameterization, not a new optimization method.

## 2. Experimental method and preregistered interpretation

We generate four mutually disjoint sets: **training** for clean checkpoint fitting; **validation** for best checkpoint, repair selection and preservation constraint; **calibration** for choosing fault magnitude before edits; **locked test** for a single final scoring pass. The dataset is Zoology MQAR (integer token key/value pairs, \`-100\` ignored labels); all target labels are unchanged after damage. No oracle clean weights or fault magnitude are made available to an editor. A single Colab notebook offers procedural blinding only—not security isolation.

### The five requested comparison families

| Arm | Editable surface | Scientific interpretation |
|---|---|---|
| **Structured CCNN edit** | Nonlinear lag kernel's five semantic coordinates | Can the structural parameterization localize the fault? |
| **CCNN tangent edit** | Five first-order lag-kernel directions at damaged checkpoint | Is any benefit only a local function-space tangent? |
| **Unrestricted FIR edit** | Five DCT coefficients; also full \(L\)-tap variant | Does flexible filter geometry dominate the same edit budget? |
| **Continued Adam** | Same five original CCNN weights; also all network weights | Is ordinary gradient optimization sufficient? |
| **No-fault placebo** | Same methods applied to the untouched clean checkpoint | Would training have improved things even without a fault? |

**No-edit** is an explicit reference. Methods share the same adaptation and validation samples; edited parameter count, optimizer steps and wall time are recorded. The full-model Adam arm is a deliberately higher-budget, stronger recovery comparator, not a capacity-matched editor.

**Fault scenarios:** (1) lag-center shift, (2) kernel-bandwidth shift, (3) query-dependent lag shift, and (4) **off-surface signal-projection corruption** as a counterexample. Calibration chooses perturbation magnitude using the affected cohort *only* and does not inspect test. A fault must produce measurable performance damage on calibration before any repair claim is interpreted.

**Primary quantities** (NLL-based; lower is better):
\[
R=\frac{L_{\mathrm{fault}}-L_{\mathrm{edit}}}{L_{\mathrm{fault}}-L_{\mathrm{clean}}},\qquad
E=(L_{\mathrm{fault}}-L_{\mathrm{edit}})-(L_{\mathrm{clean}}-L_{\mathrm{placebo}}).
\]
The recovery ratio is **undefined** when the fault is undetectable. **Source retention** is measured on unaffected short-gap queries from the locked test. Compare recovery–retention–compute **Pareto boundaries**, not only a single favorable winner.

**Advancement gate:** source-model competence, convergence evidence and actual contribution of the memory branch must be established *before* claiming a meaningful repair result. If the memory branch is unused, weight edits cannot test the hypothesized advantage; report \`unready_model\`, not a positive outcome. All faults/methods and negative outcomes remain recorded.
**Diagnostic-task refinement (17 November 2026).** The original 8,192-token PILOT was not adequately learned and the CCNN branch barely affected recall. We retain those negative/invalid observations. Before repeating edits, `PILOT` now uses the **official Zoology MQAR generator** with a 256-token vocabulary and 10,000 training examples (as in Zoology's basic-example scale), alongside a **query/key-conditioned CCNN memory modulation**. This is an architecture change, not merely more training. It is a smaller mechanism-identification experiment and must not be compared numerically with the ICLR24 Figure-2 8,192-token results. **Both clean retrieval competence and a positive memory-ablation effect remain mandatory.**

```python
# 5. Shared attention–CCNN architecture (same learned checkpoint for all editors)
# 6. Shared causal transformer and two memory mechanisms
def hist(s,lag):
    # [B,T] -> [B,T,L], newest-first; no future information.
    return F.pad(s,(lag-1,0)).unfold(1,lag,1).flip(-1)

def ode_trace(drive):
    alpha=1.0-math.exp(-8.0/drive.shape[1])
    z=torch.zeros_like(drive[:,0]);rows=[]
    for t in range(drive.shape[1]):
        z=(1-alpha)*z+alpha*torch.tanh(2.0*drive[:,t])
        rows.append(z)
    return torch.stack(rows,1)

class CCNNLag(nn.Module):
    """Original finite-lag causal convolution with 5 interpretable parameters."""
    def __init__(self):
        super().__init__()
        for k,v in (("mu0",.4),("muq",.12),("logsigma",-1.8),("freq",2.),("chirp",.5)):
            setattr(self,k,nn.Parameter(torch.tensor(v,dtype=torch.float32)))
    def parameters5(self):
        return [self.mu0,self.muq,self.logsigma,self.freq,self.chirp]
    def kernel(self,q,delta=None):
        p=self.parameters5()
        if delta is not None:p=[p[i]+delta[i] for i in range(5)]
        tau=torch.arange(LAGS,device=q.device,dtype=q.dtype).view(1,1,LAGS)/(LAGS-1)
        mu=p[0]+p[1]*q.unsqueeze(-1)
        sig=F.softplus(p[2])+.04
        raw=torch.exp(-.5*((tau-mu)/sig).square())*torch.cos(2*torch.pi*(p[3]*tau+p[4]*tau.square()))
        return F.normalize(raw,dim=-1,eps=1e-7)

class FlexibleFIR(nn.Module):
    def __init__(self):
        super().__init__()
        self.mlp=nn.Sequential(nn.Linear(5,32),nn.GELU(),nn.Linear(32,LAGS))
    def kernel(self,q):
        t=torch.linspace(0,1,q.shape[1],device=q.device,dtype=q.dtype)[None,:].expand_as(q)
        features=torch.stack([q,q.square(),torch.sin(torch.pi*q),t,t.square()],-1)
        return F.normalize(self.mlp(features),dim=-1,eps=1e-7)

class MemoryBias(nn.Module):
    def __init__(self,kind,d):
        super().__init__()
        self.kind=kind
        self.signal=nn.Linear(d,1)
        self.cond=nn.Linear(d,1)
        self.filter=CCNNLag() if kind=="ccnn" else FlexibleFIR()
        self.gain=nn.Parameter(torch.tensor(.75))
    def forward(self,h,k_override=None):
        s=self.signal(h).squeeze(-1)
        q=torch.tanh(self.cond(h)).squeeze(-1)
        weights=self.filter.kernel(q) if k_override is None else k_override(q)
        return self.gain*ode_trace((weights*hist(s,LAGS)).sum(-1))

class CausalAttentionBlock(nn.Module):
    def __init__(self,d,heads,kind):
        super().__init__()
        self.d=d;self.heads=heads
        self.ln1=nn.LayerNorm(d);self.to_qkv=nn.Linear(d,3*d)
        self.proj=nn.Linear(d,d)
        self.ln2=nn.LayerNorm(d)
        self.ff=nn.Sequential(nn.Linear(d,4*d),nn.GELU(),nn.Linear(4*d,d))
        self.memory=MemoryBias(kind,d) if kind!="attention" else None
    def forward(self,x,disable_memory=False,k_override=None,return_attention=False):
        b,t,d=x.shape
        h=self.ln1(x);q,k,v=self.to_qkv(h).chunk(3,-1)
        q=q.reshape(b,t,self.heads,d//self.heads).transpose(1,2)
        k=k.reshape(b,t,self.heads,d//self.heads).transpose(1,2)
        v=v.reshape(b,t,self.heads,d//self.heads).transpose(1,2)
        scores=(q@k.transpose(-1,-2))/math.sqrt(d//self.heads)
        if self.memory is not None and not disable_memory:
            # Experimental extension: query/key-conditioned CCNN memory modulation.
            # The earlier design added one memory-column bias independent of the
            # retrieval query; MQAR needs the current query to select matching keys.
            # Same learned attention q/k; no target labels or oracle evidence.
            bias=self.memory(h,k_override=k_override)   # [B,T_key]
            compatibility=torch.einsum(
                "bhtd,bhsd->bhts",
                F.normalize(q,dim=-1),F.normalize(k,dim=-1))
            scores=scores+bias[:,None,None,:]*compatibility
        causal=torch.ones(t,t,dtype=torch.bool,device=x.device).tril()
        scores=scores.masked_fill(~causal[None,None],torch.finfo(scores.dtype).min)
        weights=torch.softmax(scores,dim=-1)
        x=x+self.proj((weights@v).transpose(1,2).reshape(b,t,d))
        x=x+self.ff(self.ln2(x))
        return x,(weights.mean(1) if return_attention else None)

class RecallNet(nn.Module):
    def __init__(self,kind,d=128):
        super().__init__()
        assert kind in ("attention","ccnn","fir")
        self.kind=kind
        self.token=nn.Embedding(VOCAB,d)
        self.local=nn.Conv1d(d,d,3,groups=d)
        self.blocks=nn.ModuleList([CausalAttentionBlock(d,HEADS,
            kind if i==DEPTH-1 else "attention") for i in range(DEPTH)])
        self.final=nn.LayerNorm(d)
        self.head=nn.Linear(d,VOCAB,bias=False)
    def forward(self,x,labels=None,disable_memory=False,k_override=None,return_attention=False):
        z=self.token(x)
        z=z+self.local(F.pad(z.transpose(1,2),(2,0))).transpose(1,2)
        last_attention=None
        for i,layer in enumerate(self.blocks):
            z,a=layer(z,disable_memory,k_override,return_attention and i==DEPTH-1)
            if a is not None:last_attention=a
        # Model output is a function of INPUT TOKENS ONLY. Target labels cannot
        # alter attention, memory, or which positions the network predicts.
        logits=self.head(self.final(z))
        return logits,last_attention

def model_mem(model):
    return model.blocks[-1].memory
def nparams(m):
    return sum(p.numel() for p in m.parameters() if p.requires_grad)
def sha_model(m):
    h=hashlib.sha256()
    for k,v in sorted(m.state_dict().items()):
        h.update(k.encode());h.update(v.detach().cpu().contiguous().numpy().tobytes())
    return h.hexdigest()
def token_loss(model,x,y,**kw):
    logits,_=model(x,y,**kw)
    return F.cross_entropy(logits[y!=-100],y[y!=-100])
```

## 3. Clean checkpoint training: readiness is part of the scientific experiment

We train an attention–CCNN model on the pinned MQAR task and select **the best validation checkpoint**. We measure clean query-retrieval accuracy and the change when disabling the memory branch, on validation before testing. We do not inject any fault if training has not produced sufficiently capable and memory-dependent behavior.

**Configuration note:** \`SMOKE\` defaults to one tiny seed for an execution check. Set \`MODE="PILOT"\` for a meaningful exploratory study with two seeds, 6,000 independent training examples per seed, up to 40 epochs and all repair methods. \`CONFIRMATORY\` is only permitted after freezing thresholds, tuning/resource parity, and independent test governance; the code still labels it unconfirmed until those conditions are documented.

```python
# 6. Training and readiness (train and validation; locked test not accessed)
def read_metrics(model,segment,ablate=False):
    model.eval();total=correct=0;loss_sum=0.
    with torch.no_grad():
        for x,y in zip(segment["x"].split(BATCH),segment["y"].split(BATCH)):
            xx=x.to(DEVICE); yy=y.to(DEVICE)
            logits,_=model(xx,None,disable_memory=ablate)
            mask=yy!=-100
            if not mask.any():continue
            z=logits[mask]; lab=yy[mask]
            loss_sum+=float(F.cross_entropy(z,lab,reduction="sum").item())
            correct+=int((z.argmax(-1)==lab).sum().item())
            total+=int(lab.numel())
    return dict(accuracy=correct/max(total,1),nll=loss_sum/max(total,1),queries=total)
def fit_clean(seed,train,val):
    runs=[]
    for lr in LR_SWEEP:
        seeds(seed);m=RecallNet("ccnn",d=WIDTH).to(DEVICE)
        opt=torch.optim.AdamW(m.parameters(),lr=lr,weight_decay=.1)
        best=(float("inf"),None,0);since=0;history=[];start=time.perf_counter()
        for epoch in range(EPOCHS):
            m.train();perm=torch.randperm(len(train["x"]))
            for ix in perm.split(BATCH):
                xb=train["x"][ix].to(DEVICE);yb=train["y"][ix].to(DEVICE)
                opt.zero_grad()
                loss=token_loss(m,xb,yb)
                loss.backward();torch.nn.utils.clip_grad_norm_(m.parameters(),1.0);opt.step()
            score=read_metrics(m,val)
            history.append(dict(epoch=epoch+1,lr=lr,**score))
            if score["nll"]<best[0]-1e-4:
                best=(score["nll"],copy.deepcopy(m.state_dict()),epoch+1);since=0
            else:
                since+=1
            if since>=PATIENCE:break
        m.load_state_dict(best[1]);m.eval()
        runs.append(dict(model=m,best_validation_nll=best[0],
                         epoch=best[2],lr=lr,history=history,elapsed=time.perf_counter()-start))
    return min(runs,key=lambda r:r["best_validation_nll"])
datasets={};clean_models={};readiness=[];all_curves=[]
for seed in SEEDS:
    seeds(seed)
    d={name:make_data(seed,name,items) for name,items in
       (("train",TRAIN),("val",VALID),("cal",CAL),("test",TEST))}
    freeze_all_data(seed,d)
    datasets[seed]=d
    training=fit_clean(seed,d["train"][0],d["val"][0])
    clean_models[seed]=training["model"]
    clean=read_metrics(training["model"],d["val"][0])
    branch_off=read_metrics(training["model"],d["val"][0],ablate=True)
    # Readiness asks whether the network learned independent MQAR retrieval
    # and actually uses the learned CCNN memory. Not evidence of superiority:
    # an attention-only or FIR model may still be stronger at matched cost.
    # Distinguish failure to learn retrieval from failure to use memory.
    accuracy_adequate=(clean["accuracy"]>=0.50)
    memory_contributes=((branch_off["nll"]-clean["nll"])>=0.02)
    ready=(MODE!="SMOKE" and accuracy_adequate and memory_contributes)
    reason=("smoke_not_scientific" if MODE=="SMOKE" else
            "recall_not_learned_and_memory_unused" if not accuracy_adequate and not memory_contributes else
            "recall_not_learned" if not accuracy_adequate else
            "ccnn_branch_not_contributing" if not memory_contributes else
            "ready_for_fault_calibration")
    readiness.append(dict(seed=seed,ready=ready,readiness_reason=reason,
       recall_gate_passed=accuracy_adequate,memory_gate_passed=memory_contributes,
       validation_accuracy=clean["accuracy"],ablation_accuracy=branch_off["accuracy"],
       validation_nll=clean["nll"],ablation_nll=branch_off["nll"],
       memory_ablation_delta_nll=branch_off["nll"]-clean["nll"],
       best_epoch=training["epoch"],selected_lr=training["lr"],
       train_seconds=training["elapsed"],checkpoint_sha=sha_model(training["model"])))
    all_curves.extend(dict(seed=seed,**row) for row in training["history"])
    folder=OUT/"clean_checkpoints";folder.mkdir(exist_ok=True)
    torch.save({"state_dict":training["model"].state_dict(),
                "seed":seed,"architecture":"ccnn","width":WIDTH,
                "lag_length":LAGS,"clean_validation":clean,
                "sha256":sha_model(training["model"])},
               folder/f"ccnn_seed{seed}_clean.pt")
ready_df=pd.DataFrame(readiness)
pd.DataFrame(all_curves).to_csv(OUT/"clean_training_curves.csv",index=False)
ready_df.to_csv(OUT/"checkpoint_readiness.csv",index=False)
display(ready_df[["seed","ready","readiness_reason","validation_accuracy",
    "ablation_accuracy","validation_nll","memory_ablation_delta_nll",
    "best_epoch","train_seconds"]])
if not ready_df.ready.any():
    print("REPAIR HALTED BEFORE TEST: clean CCNN checkpoints did not meet the preregistered readiness gates.")
    print("Inspect clean_training_curves.csv. Do not treat method-comparison rows as repair results.")
```

## 4. Fault calibration and editing algorithms

The evaluator selects fault strength on **calibration only**. The repair algorithm sees the faulted model and independently generated adaptation examples, but not which coordinate was corrupted, the original value or the magnitude. The *same frozen faulted state* is copied into every method.

*On-manifold* faults modify semantic CCNN kernel parameters: lag center (\`mu0\`), bandwidth (\`logsigma\`) and query dependence (\`muq\`). The *off-manifold* negative control changes the memory branch's signal projection rather than the kernel itself. Its poor recovery by kernel-only methods would delimit the proposed repair space, not disprove that weight edits can work.

**Algorithm matching:** structured, tangent and five-basis FIR all optimize **five scalar edit coordinates** with the same training schedule. Sparse Adam updates the same original five CCNN parameters and exposes a critical ordinary-optimization baseline. Full FIR updates \(L\) coordinates; full Adam updates the entire model. Report cost and scope; never treat these as equal-capacity comparisons.

**Preservation rule:** candidate selection uses affected-validation improvement subject to source-validation NLL no worse than clean-source NLL plus a preregistered tolerance. Rejected candidates are scored as *no edit* on test, and rejected statuses are retained.

```python
# 7. Calibration and differential geometry of CCNN weight edits
FAULTS=("lag_center","bandwidth","query_dependence","off_surface_signal")
EDITORS=("structured5","tangent5","fir5","fir_full","adam_sparse5","adam_full")
SOURCE_PRESERVATION_REL=.10
SOURCE_PRESERVATION_ABS=.02
MIN_CALIBRATION_DAMAGE=.05
MIN_TEST_DAMAGE=.05
MIN_IMPROVEMENT=.005
FAULT_MAGNITUDES=(-8.,-4.,-2.,-1.,-.5,-.25,.25,.5,1.,2.,4.,8.)
def split_query_mask(x,y,pairs,cut):
    pos=(y!=-100).nonzero(as_tuple=False)
    b=pos[:,0];t=pos[:,1]
    keys=x[:,:2*pairs:2]
    matched=(keys[b]==x[b,t,None])
    if not bool(matched.any(-1).all()):raise ValueError("MQAR key absent")
    idx=matched.float().argmax(-1)
    is_affected=(t-2*idx)>=cut
    return (y!=-100),is_affected
def grouped_metrics(model,segment,cut):
    model.eval();total={"source":0,"affected":0};nll={"source":0.,"affected":0.}
    correct={"source":0,"affected":0}
    with torch.no_grad():
        for x,y in zip(segment["x"].split(BATCH),segment["y"].split(BATCH)):
            xx=x.to(DEVICE);yy=y.to(DEVICE)
            pred,_=model(xx,None)
            mask,affected=split_query_mask(xx,yy,segment["pairs"],cut)
            labels=yy[mask]
            loss=F.cross_entropy(pred[mask],labels,reduction="none")
            guesses=pred[mask].argmax(-1)
            for group,which in (("source",~affected),("affected",affected)):
                if not bool(which.any()):continue
                total[group]+=int(which.sum().item())
                nll[group]+=float(loss[which].sum().item())
                correct[group]+=int((guesses[which]==labels[which]).sum().item())
    return {k:dict(nll=(nll[k]/total[k] if total[k] else float("nan")),
                   accuracy=(correct[k]/total[k] if total[k] else float("nan")),
                   queries=total[k]) for k in total}
def fault_model(model,family,magnitude,seed):
    bad=copy.deepcopy(model)
    with torch.no_grad():
        mem=model_mem(bad)
        if family=="lag_center":mem.filter.mu0.add_(magnitude)
        elif family=="bandwidth":mem.filter.logsigma.add_(magnitude)
        elif family=="query_dependence":mem.filter.muq.add_(magnitude)
        elif family=="off_surface_signal":
            # Deterministic random tensor direction, unrelated to the 5-dimensional
            # kernel edit surface. Perturb the projection, NOT ground-truth labels.
            g=torch.Generator(device="cpu").manual_seed(seed+7043)
            noise=torch.randn(mem.signal.weight.shape,generator=g).to(DEVICE)
            noise=noise/(noise.norm()+1e-8)
            mem.signal.weight.add_(float(magnitude)*noise)
        else:raise ValueError(family)
    bad.eval()
    return bad
def determine_cut(segment):
    gap=segment["gap"].float()
    return max(1,int(torch.median(gap).item()))
def calibrate_damage(clean,family,cal,seed,cut):
    before=grouped_metrics(clean,cal,cut)["affected"]["nll"]
    options=[]
    for value in FAULT_MAGNITUDES:
        corrupted=fault_model(clean,family,value,seed)
        harm=grouped_metrics(corrupted,cal,cut)["affected"]["nll"]-before
        options.append(dict(harm=harm,magnitude=value,model=corrupted))
    good=[v for v in options if np.isfinite(v["harm"]) and v["harm"]>=MIN_CALIBRATION_DAMAGE]
    chosen=min(good,key=lambda v:abs(v["harm"]-.15)) if good else max(options,key=lambda v:v["harm"])
    return dict(faulted=chosen["model"],magnitude=chosen["magnitude"],
                calibration_damage=chosen["harm"],
                detectable=bool(chosen["harm"]>=MIN_CALIBRATION_DAMAGE))
def dct_basis(length,k,device):
    tau=torch.arange(length,dtype=torch.float32,device=device).view(length,1)
    j=torch.arange(k,dtype=torch.float32,device=device).view(1,k)
    return F.normalize(torch.cos(torch.pi*(tau+.5)*j/length),dim=0)
class LagKernelEditor(nn.Module):
    """One shared damaged CCNN checkpoint; learn only adapter update coordinates."""
    def __init__(self,damaged,method):
        super().__init__()
        self.net=copy.deepcopy(damaged)
        self.method=method
        for p in self.net.parameters():p.requires_grad_(False)
        count=LAGS if method=="fir_full" else 5
        self.delta=nn.Parameter(torch.zeros(count,device=DEVICE))
        self.register_buffer("basis",dct_basis(LAGS,5,DEVICE))
    def override(self,q):
        filt=model_mem(self.net).filter
        if self.method=="structured5":
            return filt.kernel(q,self.delta)
        k0=filt.kernel(q)
        if self.method=="tangent5":
            tangent=torch.zeros_like(k0)
            for i in range(5):
                probe=torch.zeros(5,device=q.device)
                probe[i]=1e-3
                direction=(filt.kernel(q,probe)-filt.kernel(q,-probe))/(2e-3)
                tangent=tangent+self.delta[i]*direction
            return F.normalize(k0+tangent,dim=-1)
        if self.method=="fir5":
            offset=self.basis@self.delta
        elif self.method=="fir_full":
            offset=self.delta
        else:raise ValueError(self.method)
        return F.normalize(k0+offset.view(1,1,-1),dim=-1)
    def forward(self,x):
        return self.net(x,None,k_override=self.override)[0]
def make_candidate(damaged,method):
    if method in ("structured5","tangent5","fir5","fir_full"):
        m=LagKernelEditor(damaged,method).to(DEVICE)
        params=[m.delta]
    else:
        m=copy.deepcopy(damaged).to(DEVICE)
        for p in m.parameters():p.requires_grad_(False)
        if method=="adam_sparse5":
            params=model_mem(m).filter.parameters5()
        elif method=="adam_full":
            params=list(m.parameters())
        else:raise ValueError(method)
        for p in params:p.requires_grad_(True)
    return m,params
def candidate_logits(candidate,x):
    if isinstance(candidate,LagKernelEditor):return candidate(x)
    return candidate(x,None)[0]
```

## 5. Validation-selected editing and sham controls

For each fault, all repair methods start from **the same corrupted checkpoint**, receive the same adaptation examples, and are selected using the same validation preservation rule. Each method is also applied to the untouched checkpoint with identical data and schedule as the **no-fault placebo**. Do not interpret a positive repair if a similarly optimized clean model gains just as much.

The underlying five-parameter structured adapter and sparse Adam share the same kernel family; similar outcomes are expected in many regimes. Any difference must be attributed to learning-rate/regularization/coordinate conditioning rather than advertised as an unknown new physical law.

**Compute disclosure:** Full-network Adam has many more editable parameters; it is the *high-capacity recovery bound* and should not be used as a nominally matched-size control.

```python
# 8. Blind editor; validation chooses edits and protects source behavior
def edit_validation(candidate,seg,cut):
    # IMPORTANT: no test data or fault oracle is visible here.
    x=seg["x"].to(DEVICE);y=seg["y"].to(DEVICE)
    with torch.no_grad():
        out=candidate_logits(candidate,x)
        mask,affected=split_query_mask(x,y,seg["pairs"],cut)
        ce=F.cross_entropy(out[mask],y[mask],reduction="none")
        src=float(ce[~affected].mean()) if bool((~affected).any()) else float("nan")
        aff=float(ce[affected].mean()) if bool(affected.any()) else float("nan")
    return src,aff
def train_editor(faulted,clean_validation_source,train,val,cut,method):
    editor,params=make_candidate(faulted,method)
    lr=(.0003 if method=="adam_full" else .02)
    opt=torch.optim.Adam(params,lr=lr)
    before_s,before_a=edit_validation(editor,val,cut)
    ceiling=clean_validation_source*(1+SOURCE_PRESERVATION_REL)+SOURCE_PRESERVATION_ABS
    best=None;start=time.perf_counter();steps=0
    for step in range(EDIT_STEPS):
        editor.train();opt.zero_grad()
        seg=train; size=len(seg["x"])
        start_ix=(step*BATCH)%size
        ix=(torch.arange(BATCH)+start_ix)%size
        x=seg["x"][ix].to(DEVICE);y=seg["y"][ix].to(DEVICE)
        out=candidate_logits(editor,x)
        mask,affected=split_query_mask(x,y,seg["pairs"],cut)
        if not bool(affected.any()):continue
        ce=F.cross_entropy(out[mask],y[mask],reduction="none")
        # Optimize only the fault-affected training population.
        loss=ce[affected].mean()
        loss.backward();torch.nn.utils.clip_grad_norm_(params,2.)
        opt.step();steps+=1
        # Limit validation work; keep the same step/evaluation policy for all methods.
        if (step+1)%5!=0 and step!=EDIT_STEPS-1:continue
        editor.eval()
        s,a=edit_validation(editor,val,cut)
        if np.isfinite(s) and np.isfinite(a) and s<=ceiling and a<before_a-MIN_IMPROVEMENT:
            if best is None or a<best[0]:
                best=(a,copy.deepcopy(editor.state_dict()),step+1)
    elapsed=time.perf_counter()-start
    count=sum(p.numel() for p in params)
    if best is None:
        return faulted,dict(accepted=False,reason="validation_rejected",
                            steps=steps,editable=count,elapsed=elapsed,
                            baseline_val_source=before_s,baseline_val_affected=before_a)
    editor.load_state_dict(best[1]);editor.eval()
    return editor,dict(accepted=True,reason="accepted",
                       steps=steps,selected_step=best[2],
                       editable=count,elapsed=elapsed,
                       baseline_val_source=before_s,baseline_val_affected=before_a)
def adapter_test_metrics(model,segment,cut):
    model.eval()
    counts={"source":0,"affected":0};losses={"source":0.,"affected":0.}
    correct={"source":0,"affected":0}
    with torch.no_grad():
        for x,y in zip(segment["x"].split(BATCH),segment["y"].split(BATCH)):
            xx=x.to(DEVICE);yy=y.to(DEVICE)
            pred=candidate_logits(model,xx)
            mask,aff=split_query_mask(xx,yy,segment["pairs"],cut)
            yy_lab=yy[mask];ce=F.cross_entropy(pred[mask],yy_lab,reduction="none")
            for name,select in (("source",~aff),("affected",aff)):
                n=int(select.sum().item())
                if n==0:continue
                counts[name]+=n;losses[name]+=float(ce[select].sum())
                correct[name]+=int((pred[mask][select].argmax(-1)==yy_lab[select]).sum())
    return {name:{"nll":losses[name]/counts[name] if counts[name] else np.nan,
                  "accuracy":correct[name]/counts[name] if counts[name] else np.nan,
                  "query_count":counts[name]} for name in counts}

# One engineering-only adapter gradient probe: not a scientific claim.
if MODE=="SMOKE":
    base=clean_models[SEEDS[0]]
    spec=datasets[SEEDS[0]]["train"][0]
    xx=spec["x"][:2].to(DEVICE);yy=spec["y"][:2].to(DEVICE)
    for method in EDITORS:
        candidate,params=make_candidate(base,method)
        pred=candidate_logits(candidate,xx)
        loss=F.cross_entropy(pred[yy!=-100],yy[yy!=-100])
        loss.backward()
        assert all(p.grad is not None and bool(torch.isfinite(p.grad).all()) for p in params),method
        assert pred.shape==(2,xx.shape[1],VOCAB)
    print("EDITOR_GRADIENT_SMOKE: all six editor surfaces differentiable")
```

## 6. Execution: four fault families × editor comparisons on the SAME trained model

The experiment writes one row for every **seed × fault × method**. If checkpoint readiness or fault calibration fails, it records a non-valid status rather than fabricating recovery. \`SMOKE\` executes a tiny independent adapter gradient check and generates provisional rows, but **never** declares repair evidence.

**Scientific decision:** Among detectable faults, compare recovery and placebo-adjusted improvement subject to source preservation. A conditional finding is useful only when we report the complete fault matrix, including cases where the flexible FIR or Adam wins.

```python
# 9. Execute repair study: no locked-test scoring when the scientific prerequisites fail
results=[];fault_log=[]
llm_fault_cases={}
for seed in SEEDS:
    d=datasets[seed]
    clean=clean_models[seed]
    cut=determine_cut(d["cal"][0])
    clean_val=grouped_metrics(clean,d["val"][0],cut)
    clean_source=clean_val["source"]["nll"]
    is_ready=bool(ready_df.loc[ready_df.seed==seed,"ready"].iloc[0])
    # The calibration split is used even when the model is unready:
    # this diagnoses whether faults affect model function at all.
    calibrated={}
    for family in FAULTS:
        fault=calibrate_damage(clean,family,d["cal"][0],seed,cut)
        calibrated[family]=fault
        llm_fault_cases[(seed,family)]=dict(faulted=fault["faulted"],
            checkpoint_ready=is_ready,
            calibration_detectable=bool(fault["detectable"]),cut=cut)
        fault_log.append(dict(seed=seed,fault=family,
            calibrated_magnitude=fault["magnitude"],
            calibration_damage=fault["calibration_damage"],
            calibration_detectable=fault["detectable"],
            clean_checkpoint_sha=sha_model(clean),
            faulted_checkpoint_sha=sha_model(fault["faulted"])))
    # If no meaningful recall or memory contribution, skip edit training AND locked test.
    # Keep one explicit missing-data row per method/fault; do not emit false zero gains.
    if not is_ready:
        for family,fault in calibrated.items():
            for method in EDITORS:
                results.append(dict(seed=seed,fault=family,method=method,
                    status="engineering_smoke" if MODE=="SMOKE" else "unready_model",
                    readiness_reason=ready_df.loc[ready_df.seed==seed,"readiness_reason"].iloc[0],
                    valid_candidate=False,checkpoint_ready=False,
                    detectable_calibration=bool(fault["detectable"]),
                    test_headroom=np.nan,recovered_fraction=np.nan,
                    test_fault_improvement=np.nan,test_placebo_improvement=np.nan,
                    placebo_adjusted_gain=np.nan,clean_source_nll=np.nan,
                    source_after_nll=np.nan,source_drift=np.nan,
                    source_preserved=False,affected_clean_accuracy=np.nan,
                    affected_faulted_accuracy=np.nan,affected_repaired_accuracy=np.nan,
                    accepted_on_validation=False,placebo_accepted_on_validation=False,
                    editable_scalars=(sum(p.numel() for p in clean.parameters())
                        if method=="adam_full" else LAGS if method=="fir_full" else 5),
                    repair_optimizer_steps=0,repair_seconds=0.,
                    faulted_checkpoint_sha=sha_model(fault["faulted"])))
        continue
    # Run the zero-fault placebo once per editor, then score only after selection.
    placebo_cache={}
    for method in EDITORS:
        sham,sham_decision=train_editor(clean,clean_source,
                            d["train"][0],d["val"][0],cut,method)
        placebo_cache[method]=(sham_decision,sham)
    # Locked test is not involved in any checkpoint, fault or edit selection.
    clean_locked=adapter_test_metrics(clean,d["test"][0],cut)
    for family,fault in calibrated.items():
        calibration_detectable=bool(fault["detectable"])
        damaged_locked=(adapter_test_metrics(fault["faulted"],d["test"][0],cut)
                        if calibration_detectable else None)
        for method in EDITORS:
            if calibration_detectable:
                edited,decision=train_editor(fault["faulted"],clean_source,
                                  d["train"][0],d["val"][0],cut,method)
                sham_decision,sham=placebo_cache[method]
                em=adapter_test_metrics(edited,d["test"][0],cut)
                pm=adapter_test_metrics(sham,d["test"][0],cut)
                cm=clean_locked
                fm=damaged_locked
                headroom=fm["affected"]["nll"]-cm["affected"]["nll"]
                improvement=fm["affected"]["nll"]-em["affected"]["nll"]
                placebo=cm["affected"]["nll"]-pm["affected"]["nll"]
                preserved=bool(em["source"]["nll"]<=
                   cm["source"]["nll"]*(1+SOURCE_PRESERVATION_REL)+SOURCE_PRESERVATION_ABS)
                valid=bool(headroom>=MIN_TEST_DAMAGE and decision["accepted"]
                           and preserved and improvement>placebo)
                status=("undetectable_test_fault" if headroom<MIN_TEST_DAMAGE else
                        "validation_rejected" if not decision["accepted"] else
                        "source_regression" if not preserved else
                        "not_fault_specific" if improvement<=placebo else
                        "fault_specific_candidate")
                source_drift=em["source"]["nll"]-cm["source"]["nll"]
                source_clean=cm["source"]["nll"]; source_after=em["source"]["nll"]
                clean_acc=cm["affected"]["accuracy"]
                fault_acc=fm["affected"]["accuracy"]
                edit_acc=em["affected"]["accuracy"]
            else:
                decision=dict(accepted=False,steps=0,editable=(
                     sum(p.numel() for p in clean.parameters())
                     if method=="adam_full" else LAGS if method=="fir_full" else 5),
                     elapsed=0.)
                sham_decision=dict(accepted=False)
                headroom=improvement=placebo=source_drift=np.nan
                source_clean=source_after=clean_acc=fault_acc=edit_acc=np.nan
                valid=False;preserved=False
                status="undetectable_calibration_fault"
            results.append(dict(seed=seed,fault=family,method=method,
                status=status,readiness_reason="ready_for_fault_calibration",
                valid_candidate=valid,checkpoint_ready=True,
                detectable_calibration=calibration_detectable,
                test_headroom=headroom,
                recovered_fraction=(improvement/headroom if
                    np.isfinite(headroom) and headroom>=MIN_TEST_DAMAGE else np.nan),
                test_fault_improvement=improvement,
                test_placebo_improvement=placebo,
                placebo_adjusted_gain=(improvement-placebo if
                     np.isfinite(improvement) and np.isfinite(placebo) else np.nan),
                clean_source_nll=source_clean,source_after_nll=source_after,
                source_drift=source_drift,source_preserved=preserved,
                affected_clean_accuracy=clean_acc,
                affected_faulted_accuracy=fault_acc,
                affected_repaired_accuracy=edit_acc,
                accepted_on_validation=bool(decision["accepted"]),
                placebo_accepted_on_validation=bool(sham_decision["accepted"]),
                editable_scalars=int(decision["editable"]),
                repair_optimizer_steps=int(decision["steps"]),
                repair_seconds=float(decision["elapsed"]),
                faulted_checkpoint_sha=sha_model(fault["faulted"])))
outcomes=pd.DataFrame(results)
faults=pd.DataFrame(fault_log)
outcomes.to_csv(OUT/"all_fault_method_trials.csv",index=False)
faults.to_csv(OUT/"evaluator_fault_ledger_DO_NOT_GIVE_EDITOR.csv",index=False)
print("CHECKPOINT_READINESS")
display(ready_df[["seed","ready","readiness_reason","validation_accuracy",
                  "memory_ablation_delta_nll","best_epoch"]])
print("FAULT_SENSITIVITY_CALIBRATION (independent of test)")
display(faults[["seed","fault","calibration_damage","calibration_detectable"]])
print("REPAIR_STATUS_COUNTS")
display(outcomes.status.value_counts(dropna=False).rename_axis("status").reset_index(name="count"))
valid=outcomes[outcomes["status"]=="fault_specific_candidate"]
if valid.empty:
    print("NO VALID REPAIR COMPARISON. Models must first learn recall and use their CCNN branch.")
else:
    display(valid.groupby(["fault","method"],as_index=False).agg(
        valid_trials=("seed","size"),
        placebo_adjusted_gain=("placebo_adjusted_gain","mean"),
        source_drift=("source_drift","mean")))
print("MODE",MODE,"READY_MODELS",int(ready_df.ready.sum()),
      "CALIBRATED_FAULTS",int(faults.calibration_detectable.sum()))
```

## 7. LLM-guided repair: does CCNN prepare weights for an LLM?

**Research extension H-LLM:** the five interpretable lag-kernel coordinates and their response curves may improve an LLM's ability to propose small, localized repairs compared with giving it exactly the same damaged coordinates as anonymous numbers. Test correct semantics, anonymous weights, and mismatched semantic names. This is a conjecture to test, not an established CCNN advantage.

Recent *Sparks of Weight-Space Programming* demonstrations show instances of LLM-written model weights; *Can an LLM Actually Reason and Program about Neural Network's Weights?* (arXiv:2611.08070) and *WeightBench* (arXiv:2611.18553) motivate strong controls for hallucinated weight reasoning, false semantic priors, and impossible edits.

- llm_raw: five numeric coordinates plus the same validation finite-difference probes
- llm_ccnn: actual causal lag-center, conditional shift, bandwidth, frequency and chirp semantics plus sampled kernel responses
- llm_permuted: intentionally shuffled coordinate names, with numeric values and probes unchanged

The LLM selects a vector of five scalar weight changes, or explicitly abstains. A numerical validator may only choose a predefined scale of the LLM's vector or no edit; no Adam step is disguised as LLM reasoning. Same train/validation/calibration/locked-test boundaries, no clean checkpoint or injected fault identity in prompts. An independent LLM no-fault placebo proposal tests whether its apparent benefit is continued model improvement.

**Local Hugging Face execution:** Install `transformers accelerate bitsandbytes` in Colab and set `LLM_MODE="hf_local"`. The notebook loads [Qwen3-4B-Instruct-2507](https://huggingface.co/Qwen/Qwen3-4B-Instruct-2507) (Apache-2.0) with NF4 4-bit quantization; no proprietary API or OpenAI key is used. A GPU is required. The default `off` mode makes no LLM call and invents no LLM outputs. Run after sufficient clean training, fault calibration and checkpoint readiness. Compare actual generated edits; a malformed response is reported as an error, never converted to a synthetic repair.

**References:** [Sparks of Weight-Space Programming](https://rikiyat.github.io/blog/weight-space-programming/); [weight-reasoning audit](https://arxiv.org/abs/2611.08070); [WeightBench](https://arxiv.org/abs/2611.18553).
**Information-matched control:** Each prompt arm receives the same five numerical values, local validation sensitivities, and sampled kernel response. Only the coordinate naming/semantics are altered. The permuted-name arm deliberately injects wrong semantic labels as a falsification control.


```python
# 10. Real LLM proposal experiment, optional; no invented or heuristic LLM outputs
LLM_MODE="off"        # "off" (safe default) | "hf_local" (real model on Colab GPU)
LLM_MODEL="Qwen/Qwen3-4B-Instruct-2507"  # Apache-2.0; Hugging Face, no API key
LLM_4BIT=True         # quantized local inference for a T4/L4 GPU
LLM_MAX_NEW_TOKENS=220
LLM_DELTA_LIMIT=8.0
LLM_VARIANTS=("llm_raw","llm_ccnn","llm_permuted")
assert LLM_MODE in ("off","hf_local")

def llm_weights_prompt(model,validation,cut,variant):
    vals=[float(p.detach().cpu()) for p in model_mem(model).filter.parameters5()]
    meaningful=["delay_center","query_shift","log_bandwidth","frequency","chirp"]
    names=(meaningful if variant=="llm_ccnn" else
           meaningful[2:]+meaningful[:2] if variant=="llm_permuted" else
           [f"coordinate_{i}" for i in range(5)])
    slopes=[]
    # Identical validation-only local sensitivity probes in all prompt conditions.
    for i in range(5):
        dp=torch.zeros(5,device=DEVICE);dp[i]=.03
        positive=LagKernelEditor(model,"structured5").to(DEVICE)
        negative=LagKernelEditor(model,"structured5").to(DEVICE)
        with torch.no_grad():
            positive.delta.copy_(dp);negative.delta.copy_(-dp)
        sp,ap=edit_validation(positive,validation,cut)
        sm,am=edit_validation(negative,validation,cut)
        slopes.append({"index":i,"affected_slope":round((ap-am)/.06,5),
                       "source_slope":round((sp-sm)/.06,5)})
    observed=grouped_metrics(model,validation,cut)
    view={"coords":[{"index":i,"name":names[i],"value":round(vals[i],5)}
                     for i in range(5)],
          "slopes":slopes,
          "affected_nll":observed["affected"]["nll"],
          "source_nll":observed["source"]["nll"],
          "required_output":{"abstain":False,"deltas":[0,0,0,0,0],"reason":"..."}}
    # All variants receive the same numerical information, including response curves.
    # Only the semantic mapping changes; avoid confounding interpretation with extra data.
    with torch.no_grad():
        q=torch.tensor([[-.75,0.,.75]],device=DEVICE)
        k=model_mem(model).filter.kernel(q)[0]
        idx=torch.linspace(0,LAGS-1,8).long()
        view["kernel_samples"]=[[round(float(z),4) for z in row] for row in k[:,idx].cpu()]
    if variant=="llm_ccnn":
        view["interpretation"]=(
          "The causal lag filter has delay center, query-dependent shift, log bandwidth, "
          "oscillation frequency and chirp. Its output drives a leaky state modifying "
          "the attention scores. Propose a small weight correction; source recall must persist.")
    elif variant=="llm_permuted":
        view["interpretation"]="Coordinate names are potentially unreliable. Infer from numeric probes."
    else:
        view["interpretation"]="Five editable coefficients; architecture semantics withheld."
    return ("You are a weight-space repair proposer. Choose exactly five bounded scalar "
            "deltas IN INDEX ORDER (0..4), or abstain. Do not assume an oracle fault "
            "identity, do not request clean reference weights, do not use test results. "
            "Output JSON only.\\n"+json.dumps(view,sort_keys=True))

def llm_client():
    """Load a real open-source local Hugging Face model; no API tokens."""
    if LLM_MODE!="hf_local":return None
    if not torch.cuda.is_available():
        raise RuntimeError("HF local 4B LLM requires a GPU: Runtime > Change runtime type > T4 GPU.")
    try:
        from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
    except ImportError as ex:
        raise RuntimeError("Install in an earlier Colab cell: %pip install -q -U transformers accelerate bitsandbytes") from ex
    tokenizer=AutoTokenizer.from_pretrained(LLM_MODEL,trust_remote_code=False)
    params=dict(device_map="auto",trust_remote_code=False,torch_dtype=torch.float16)
    if LLM_4BIT:
        params["quantization_config"]=BitsAndBytesConfig(
            load_in_4bit=True,bnb_4bit_quant_type="nf4",
            bnb_4bit_compute_dtype=torch.float16,bnb_4bit_use_double_quant=True)
    model=AutoModelForCausalLM.from_pretrained(LLM_MODEL,**params)
    model.eval()
    print("LOCAL_HF_MODEL_LOADED",LLM_MODEL,"four_bit",LLM_4BIT,
          "parameters",sum(p.numel() for p in model.parameters()))
    return {"model":model,"tokenizer":tokenizer}

def llm_one_proposal(client,prompt):
    """Return only a REAL generated edit; preserve invalid JSON as an error."""
    import re
    from types import SimpleNamespace
    model=client["model"];tok=client["tokenizer"]
    started=time.perf_counter()
    messages=[
      {"role":"system","content":
       "You are a model-weight repair proposer. Respond only with one JSON object with keys abstain, deltas and reason. No markdown."},
      {"role":"user","content":prompt}]
    rendered=tok.apply_chat_template(
        messages,tokenize=False,add_generation_prompt=True)
    inputs=tok(rendered,return_tensors="pt")
    target_device=next(model.parameters()).device
    inputs={k:v.to(target_device) for k,v in inputs.items()}
    with torch.inference_mode():
        output=model.generate(
            **inputs,max_new_tokens=LLM_MAX_NEW_TOKENS,
            do_sample=False,pad_token_id=tok.eos_token_id)
    new_ids=output[0][inputs["input_ids"].shape[-1]:]
    raw=tok.decode(new_ids,skip_special_tokens=True).strip()
    # Accept a JSON object surrounded by harmless prose but never fabricate
    # a direction if the local model fails to return parseable structured output.
    match=re.search(r"\{[\s\S]*\}",raw)
    if match is None:
        raise ValueError("Local HF model did not return a JSON object: "+raw[:150])
    obj=json.loads(match.group(0))
    if not isinstance(obj,dict) or not isinstance(obj.get("abstain"),bool):
        raise ValueError("LLM output requires boolean abstain")
    usage=SimpleNamespace(prompt_tokens=int(inputs["input_ids"].shape[-1]),
                          completion_tokens=int(new_ids.numel()))
    elapsed=time.perf_counter()-started
    delta=obj.get("deltas")
    if obj["abstain"]:return None,elapsed,usage
    if not isinstance(delta,list) or len(delta)!=5 or any(
        not isinstance(v,(int,float)) or isinstance(v,bool) or
        not np.isfinite(v) or abs(v)>LLM_DELTA_LIMIT for v in delta):
        raise ValueError("Invalid bounded five-coordinate proposal: "+raw[:150])
    return torch.tensor(delta,device=DEVICE,dtype=torch.float32),elapsed,usage

def validate_llm_vector(model,val,clean_val_source,cut,delta):
    """Fixed line scales only; no gradient optimizer or arbitrary rewrite."""
    baseline=edit_validation(model,val,cut)[1]
    ceiling=clean_val_source*(1+SOURCE_PRESERVATION_REL)+SOURCE_PRESERVATION_ABS
    winner=None
    for scale in (.25,.5,1.):
        edit=LagKernelEditor(model,"structured5").to(DEVICE)
        with torch.no_grad():edit.delta.copy_(delta*scale)
        src,aff=edit_validation(edit,val,cut)
        if np.isfinite(src) and np.isfinite(aff) and src<=ceiling and aff<baseline-MIN_IMPROVEMENT:
            if winner is None or aff<winner[0]:winner=(aff,edit,scale)
    return (winner[1],True,winner[2]) if winner else (model,False,None)

llm_rows=[];llm_prompts=[]
cases=[(seed,family,v) for (seed,family),v in sorted(llm_fault_cases.items())
       if v["checkpoint_ready"] and v["calibration_detectable"]]
client=llm_client() if cases and LLM_MODE=="hf_local" else None
for seed,family,item in cases:
    data=datasets[seed];clean=clean_models[seed];bad=item["faulted"];cut=item["cut"]
    source_val=grouped_metrics(clean,data["val"][0],cut)["source"]["nll"]
    for variant in LLM_VARIANTS:
        prompt=llm_weights_prompt(bad,data["val"][0],cut,variant)
        llm_prompts.append({"seed":seed,"variant":variant,"prompt":prompt,
                            "damaged_sha256":sha_model(bad)})
        row=dict(seed=seed,fault=family,variant=variant,
                 status="not_run",real_llm=False,
                 placebo_adjusted_gain=np.nan,source_drift=np.nan,
                 recovered_fraction=np.nan,repair_accepted=False,
                 elapsed_seconds=np.nan,prompt_tokens=np.nan,completion_tokens=np.nan)
        if client is not None:
            try:
                delta,sec,usage=llm_one_proposal(client,prompt)
                row.update(real_llm=True,elapsed_seconds=sec,
                           prompt_tokens=getattr(usage,"prompt_tokens",np.nan),
                           completion_tokens=getattr(usage,"completion_tokens",np.nan))
                if delta is None:
                    row["status"]="llm_abstained"
                else:
                    proposal,accepted,scale=validate_llm_vector(
                        bad,data["val"][0],source_val,cut,delta)
                    row["repair_accepted"]=accepted
                    # A separate real LLM call on clean weights provides the no-fault placebo.
                    sham_delta,sham_sec,sham_usage=llm_one_proposal(
                        client,llm_weights_prompt(clean,data["val"][0],cut,variant))
                    sham=clean
                    if sham_delta is not None:
                        sham,_,_=validate_llm_vector(clean,data["val"][0],source_val,cut,sham_delta)
                    cm=adapter_test_metrics(clean,data["test"][0],cut)
                    fm=adapter_test_metrics(bad,data["test"][0],cut)
                    pm=adapter_test_metrics(proposal,data["test"][0],cut)
                    sm=adapter_test_metrics(sham,data["test"][0],cut)
                    headroom=fm["affected"]["nll"]-cm["affected"]["nll"]
                    gain=fm["affected"]["nll"]-pm["affected"]["nll"]
                    sham_gain=cm["affected"]["nll"]-sm["affected"]["nll"]
                    drift=pm["source"]["nll"]-cm["source"]["nll"]
                    row.update(placebo_adjusted_gain=gain-sham_gain,
                        source_drift=drift,
                        recovered_fraction=(gain/headroom if headroom>=MIN_TEST_DAMAGE else np.nan),
                        status=("conditional_repair_candidate" if accepted and
                           gain>sham_gain and headroom>=MIN_TEST_DAMAGE and
                           drift<=cm["source"]["nll"]*SOURCE_PRESERVATION_REL+
                           SOURCE_PRESERVATION_ABS else "not_fault_specific"),
                        elapsed_seconds=sec+sham_sec)
            except Exception as e:
                row["status"]="llm_error";row["error"]=str(e)[:250]
        llm_rows.append(row)
llm_df=pd.DataFrame(llm_rows)
llm_df.to_csv(OUT/"llm_repair_comparison.csv",index=False)
(OUT/"llm_repair_prompts.jsonl").write_text(
    "\n".join(json.dumps(v) for v in llm_prompts))
print("LLM_MODE",LLM_MODE,"eligible faults",len(cases),
      "real LLM proposals",sum(v["real_llm"] for v in llm_rows))
if not llm_df.empty:display(llm_df)
else:print("No ready and calibration-detectable fault: LLM comparison correctly not run.")
```

## 7. Results: mechanism-specific repair boundary (generated by execution)

Graphs are meaningful only for truly ready checkpoints with detectable faults. A synthetic fault is a controlled laboratory test, not yet evidence that an editor can diagnose unknown natural failures.

## 8. Discussion: write findings from evidence, not optimism

After execution, report a matrix of **fault family → which editor wins under the preservation constraint → why the representation helped/failed → boundary/next experiment**. Examples below are *hypothesized interpretations*, not observed claims:

| Empirical pattern (to be checked) | Mechanistic interpretation | Follow-up scientific question |
|---|---|---|
| Structured nonlinear edit wins on large on-manifold faults but not tiny ones | Curvature of lag-parameter manifold matters beyond first-order tangent | Is there an identifiable trust region or conditioning criterion? |
| Structured and sparse Adam on same five parameters tie | The benefit is the **editable parameterization**, not a new optimizer | Can structure improve edit selection when coordinate identity is unknown? |
| FIR correction wins under off-manifold faults | CCNN kernel is not expressive enough for the defect | Could a gated residual repair handle such faults without source regression? |
| Full Adam recovers more but degrades unaffected recall | Locality–flexibility trade-off | Is projection onto preserved-capability directions worth the cost? |
| Zero-fault placebo improves similarly to faulty edit | Apparent repair is continued training | Do cleaner converged checkpoints or stricter sham tests remove this confound? |
| All edits ineffective because CCNN memory contributes nothing | The architecture is not using the proposed structural branch | Revisit attention coupling; **do not fabricate a repair benefit** |

### Limitations and next research direction
This initial study tests engineered model-weight defects in one memory architecture. Before claiming general weight-space programming, extend to naturally arising drift, blind fault localization, multi-model repair, sequential edits and competitive second-order preservation controls. The rationale for those extensions must derive from the observed fault-level boundary, not from generic future-work boilerplate.
